```python
import numpy as np
product = np.array([10,21,45,65,66,75,23,14,25,36,47,58])
print (product)
```

```python
discounted_prices = np.where(
    product < 300, product * 0.95,
    np.where(product <= 1000, product * 0.90, product * 0.85)
)

print(discounted_prices)
```

```python
bulk = np.where( product > 50 , product * 0.4 , product)
print(bulk)
```
