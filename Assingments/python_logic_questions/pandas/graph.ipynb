```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([1,2,3])
y = [2,10,5,4,7,9]
```

```python
unifrom_dis = np.random.uniform(13,23,10)
normal_dis = np.random.normal(10,50,13)
plt.title("random")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)

plt.bar(x,y,label ="xy plot")
plt.plot(unifrom_dis, label ="uniform_distribution")
plt.plot(normal_dis, label ="normal_distribution")
plt.legend()
plt.bar(x,y)
plt.show()

plt.savefig("random.png")
```

```python
colors = ["red","green","black"]
labels = ["age","gender","krish"]
plt.pie(x,colors=colors, labels=labels)
plt.show()
```
